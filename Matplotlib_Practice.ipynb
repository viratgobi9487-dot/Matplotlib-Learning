```python
#3  customizing plots (styling & formating) 
# "-"    # Solid
# "--"   # Dashed
# ":"    # Dotted
# "-."   # Dash-dot


x = ["tamil", "english","maths","science"]
y = [99,67,89,93]
plt.plot(x,y, linestyle= ":", marker="o", label="mark  list", color ="yellow")
plt.xlabel("language")
plt.ylabel("marks")
plt.legend(loc="upper right")
plt.show()
```

```python
# #using differt colour maps

# | Color Map   | Description                     |
# | ----------- | ------------------------------- |
# | `"viridis"` | Blue → Green → Yellow (Default) |
# | `"plasma"`  | Purple → Yellow                 |
# | `"inferno"` | Black → Orange → Yellow         |
# | `"magma"`   | Black → Purple → Orange         |
# | `"cividis"` | Blue → Yellow                   |
# | `"cool"`    | Cyan → Magenta                  |
# | `"hot"`     | Black → Red → Yellow            |
```

```python
#scatter
import matplotlib.pyplot as plt

x = ["Tamil", "English", "Maths", "Science"]
y = [99, 67, 89, 93]

colors = [20, 40, 60, 80]   # Color values

plt.plot(x, y,
         linestyle=":",
         marker="o",
         label="Mark List",
         color="yellow")

plt.scatter(x, y,
            c=colors,
            cmap="plasma",
            s=500)

plt.xlabel("Language")
plt.ylabel("Marks")
plt.legend(loc="upper right")

plt.colorbar()   # Optional: shows color scale

plt.show()
```

```python
#adding annotation text 
import matplotlib.pyplot as plt

x = ["Tamil", "English", "Maths", "Science"]
y = [99, 67, 89, 93]

plt.plot(x, y, marker="o")

plt.annotate("Highest", xy=("Tamil", 99))

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
 
```

```python
import matplotlib.pyplot as plt

# Employee Sales Data
employees = ["Arun", "Bala", "Charan", "Deepa", "Eswar"]
sales = [12, 18, 15, 22, 17]

# Figure
# plt.figure(figsize=(8,5))

# Plot
plt.plot(
    employees,
    sales,
    color="green",
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Monthly Sales"
)

# Title
plt.title("Employee Monthly Sales Report")

# Axis Labels
plt.xlabel("Employee Name")
plt.ylabel("Sales (Lakhs)")

# Grid
plt.grid(True, linestyle=":")

# Legend
plt.legend(loc="upper left")

# Show Graph
plt.show()
```
